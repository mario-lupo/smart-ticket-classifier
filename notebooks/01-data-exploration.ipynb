```python
print("="*80)
print("EXPLORATORY DATA ANALYSIS")
print("="*80)

print("Osserviamo prima la forma del dataset, i nomi delle variabili e alcune righe di esempio per capire di cosa si tratta.")
import pandas as pd # pandas serve per fare operazioni sui dataframe e serie temporali
import numpy as np # invece numpy serve per fare operazioni matematiche su array multidimensionali
import matplotlib.pyplot as plt # matplotlib serve per fare grafici
import seaborn as sns # seaborn serve per fare grafici statistici

sns.set_style("whitegrid") # impostiamo lo stile dei grafici con seaborn, in questo caso "whitegrid" cioè sfondo bianco con griglia
plt.rcParams['figure.figsize'] = (12, 6) # impostiamo la dimensione di default dei grafici
# sns e plt lavorano insieme per fare grafici più belli e personalizzati

# Carica dataset
df = pd.read_csv('../data/issues.csv', engine='python', encoding='latin1', sep=';') # carichiamo il dataset in un dataframe pandas

print("Dataset caricato! 😁")

print("\n1️⃣OVERVIEW GENERALE:")
print(f"Il numero di righe è {df.shape[0]} e di colonne è {df.shape[1]}\n")
print(f"Abbiamo usato il metodo df.shape per ottenere una tupla con il numero di righe e colonne del dataframe.")
print("Così abbiamo la carta d'identità del nostro dataset!\n")
print(f"I nomi delle variabili (colonne) sono: \n\n{df.columns}")
# perché df.columns e non df.columns() ? Perché columns è un attributo del dataframe, non una funzione
# cosa significa questo in termini semplici? Che non dobbiamo mettere le parentesi tonde quando lo usiamo perché non è una funzione che deve essere eseguita, ma un valore che deve essere letto.
# quindi è come fare df["colonna"], ma in questo caso stiamo leggendo tutti i nomi delle colonne.
```

```python
print("Abbiamo visto, attraverso df.info, che ci sono vari dati mancanti.")
print("con isnull() otteniamo un dataframe di booleani True/False per ogni cella del dataframe originale, dove True indica un valore nullo.")
print("ma per calcolare la percentuale di valori nulli per colonna, dobbiamo prima sommare i valori True (che sono considerati come 1) per ogni colonna con sum(), ottenendo così il numero di valori nulli per colonna")
print("Usiamo df.isnull().sum() per contare quanti valori nulli ci sono in ogni colonna, e poi calcoliamo la percentuale di valori nulli per colonna.")
print("poi dividiamo per il numero totale di righe (len(df)) (quali righe? Tutte le righe di quella colonna, quindi se ho 50 valori nulli su 1000 righe in quella colonna , 50/1000 = 0.05)")
null_pct = df.isnull().sum()
print(f"\nValori nulli per colonna:\n{null_pct}\n\n")
print("Voglio i valori in ordine discendente, uso sort_values() applicato alla serie null_pct, e imposto ascending=False per avere l'ordine discendente.")
null_pct = null_pct.sort_values(ascending=False)
for i in null_pct.index:
    if null_pct[i] > 0:
        print(f"La colonna '{i}' ha {null_pct[i]} valori nulli, che rappresentano il {null_pct[i]/len(df)*100:.2f}% del totale.")
# perché ho fatto .index? Perché null_pct è una serie pandas, e lo uso per iterare sulle sue etichette (i nomi delle colonne)
# infatti .index mi restituisce un oggetto Index con i nomi delle colonne della serie
# quindi in questo caso sarà equivalente a fare for i in null_pct.keys()
```
